# Fonction quadratique

Une fonction quadratique s'écrit sous la forme :

$$f(x) = ax^2 + bx + c$$

avec $a \neq 0$.

- La courbe est une parabole.
- Si $a > 0$, la parabole est ouverte vers le haut.
- Si $a < 0$, la parabole est ouverte vers le bas.
- L'axe de symétrie est donné par $x = -\frac{b}{2a}$.
- Le sommet a pour abscisse $x_s = -\frac{b}{2a}$.

```python
import numpy as np
import matplotlib.pyplot as plt

def fonction_quadratique(x, a, b, c):
    return a * x**2 + b * x + c


def racines_quadratique(a, b, c):
    discriminant = b**2 - 4 * a * c
    if discriminant < 0:
        return []
    elif discriminant == 0:
        return [-b / (2 * a)]
    else:
        return [(-b - np.sqrt(discriminant)) / (2 * a), (-b + np.sqrt(discriminant)) / (2 * a)]


def afficher_fonction_quadratique(a, b, c, x_min=-10, x_max=10, nb_points=400):
    if a == 0:
        raise ValueError("Le paramètre a doit être différent de 0.")

    x = np.linspace(x_min, x_max, nb_points)
    y = fonction_quadratique(x, a, b, c)

    x_s = -b / (2 * a)
    y_s = fonction_quadratique(x_s, a, b, c)
    racines = racines_quadratique(a, b, c)

    if racines:
        print("Racines :", racines)
    else:
        print("Aucune racine réelle.")

    plt.figure(figsize=(8, 4))
    plt.plot(x, y, label=f"f(x) = {a}x² + {b}x + {c}")
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    plt.scatter([x_s], [y_s], color='red', zorder=3, label=f"Sommet ({x_s:.2f}, {y_s:.2f})")

    if racines:
        for r in racines:
            plt.scatter([r], [0], color='green', zorder=4, label=f"y=0 en x={r:.2f}")

    plt.title('Fonction quadratique')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()


# Exemple d'utilisation
afficher_fonction_quadratique(1, -3, 2)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def tracer_valeur_absolue(a, b, c, d, xmin=-10, xmax=10):
    """
    Trace f(x) = a|bx + c| + d
    et indique le sommet et les zéros.
    """

    if b == 0:
        raise ValueError("b doit être différent de 0.")

    # Fonction
    def f(x):
        return a * np.abs(b*x + c) + d

    # -----------------------------
    # 1. Sommet
    # -----------------------------
    xs = -c / b
    ys = d

    # -----------------------------
    # 2. Recherche des zéros
    # a|bx+c| + d = 0
    # |bx+c| = -d/a
    # -----------------------------
    zeros = []

    if a != 0 and -d/a >= 0:
        k = -d/a

        x1 = ( k - c) / b
        x2 = (-k - c) / b

        zeros = [x1]

        # Évite de mettre deux fois le même zéro
        if not np.isclose(x1, x2):
            zeros.append(x2)

    # -----------------------------
    # 3. Tracé
    # -----------------------------
    x = np.linspace(xmin, xmax, 1000)
    y = f(x)

    plt.figure(figsize=(9, 6))

    plt.plot(x, y, label=f"f(x) = {a}|{b}x + {c}| + {d}")

    # Axes
    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)

    # Sommet
    plt.scatter(xs, ys, s=80, zorder=5)
    plt.annotate(
        f"Sommet ({xs:.2f}, {ys:.2f})",
        (xs, ys),
        xytext=(10, 15),
        textcoords="offset points"
    )

    # Zéros
    for z in zeros:
        plt.scatter(z, 0, s=80, zorder=5)
        plt.annotate(
            f"Zéro ({z:.2f}, 0)",
            (z, 0),
            xytext=(10, -20),
            textcoords="offset points"
        )

    plt.grid(True, alpha=0.3)
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.legend()
    plt.title("Fonction valeur absolue")

    plt.show()

    # Résultats numériques
    print(f"Sommet : ({xs:.2f}, {ys:.2f})")

    if zeros:
        print("Zéro(s) :", [round(z, 4) for z in zeros])
    else:
        print("Aucun zéro réel")
```

```python
tracer_valeur_absolue(1, -4, 80,-6, xmin=-10, xmax=50)
```

```python
import numpy as np
import matplotlib.pyplot as plt


def tracer_inegalite_valeur_absolue(
        a, b, c, d,
        m, p,
        signe="<=",
        xmin=-15,
        xmax=15):

    """
    Compare :

        f(x) = a|bx + c| + d
        g(x) = mx + p

    signe : "<=", ">=", "<" ou ">"

    Trace les deux fonctions, le sommet de f,
    les intersections et indique l'ensemble solution.
    """

    # -------------------------
    # Fonctions
    # -------------------------

    def f(x):
        return a * np.abs(b*x + c) + d

    def g(x):
        return m*x + p

    # -------------------------
    # Sommet de f
    # -------------------------

    xs = -c / b
    ys = d

    # -------------------------
    # Domaine pour le graphique
    # -------------------------

    x = np.linspace(xmin, xmax, 10000)

    yf = f(x)
    yg = g(x)

    # -------------------------
    # Recherche des intersections
    # f(x) - g(x) = 0
    # -------------------------

    h = yf - yg

    indices = np.where(np.sign(h[:-1]) != np.sign(h[1:]))[0]

    intersections = []

    for i in indices:

        # interpolation linéaire pour améliorer
        # la précision de l'intersection

        x1, x2 = x[i], x[i+1]
        h1, h2 = h[i], h[i+1]

        xi = x1 - h1 * (x2 - x1) / (h2 - h1)

        # éviter les doublons
        if not intersections or not np.isclose(
                xi, intersections[-1], atol=1e-3):
            intersections.append(xi)

    # -------------------------
    # Déterminer où l'inégalité
    # est vraie
    # -------------------------

    if signe == "<=":
        solution = yf <= yg

    elif signe == "<":
        solution = yf < yg

    elif signe == ">=":
        solution = yf >= yg

    elif signe == ">":
        solution = yf > yg

    else:
        raise ValueError("Signe invalide")

    # -------------------------
    # Graphique
    # -------------------------

    plt.figure(figsize=(10, 7))

    plt.plot(
        x, yf,
        label=f"f(x) = {a}|{b}x + {c}| + {d}",
        linewidth=2
    )

    plt.plot(
        x, yg,
        label=f"g(x) = {m}x + {p}",
        linewidth=2
    )

    # axes

    plt.axhline(0, linewidth=1)
    plt.axvline(0, linewidth=1)

    # -------------------------
    # Sommet
    # -------------------------

    plt.scatter(xs, ys, s=80, zorder=5)

    plt.annotate(
        f"Sommet ({xs:.2f}, {ys:.2f})",
        (xs, ys),
        xytext=(10, 15),
        textcoords="offset points"
    )

    # -------------------------
    # Intersections
    # -------------------------

    for xi in intersections:

        yi = f(xi)

        plt.scatter(xi, yi, s=80, zorder=5)

        plt.annotate(
            f"({xi:.2f}, {yi:.2f})",
            (xi, yi),
            xytext=(10, 10),
            textcoords="offset points"
        )

    # -------------------------
    # Zone solution sur axe X
    # -------------------------

    ymin = min(yf.min(), yg.min())

    plt.fill_between(
        x,
        ymin,
        ymin + (max(yf.max(), yg.max()) - ymin) * 0.04,
        where=solution,
        alpha=0.4,
        label="Ensemble solution"
    )

    plt.grid(True, alpha=0.3)

    plt.xlabel("x")
    plt.ylabel("y")

    plt.title(f"Résolution graphique : f(x) {signe} g(x)")

    plt.legend()

    plt.show()

    # -------------------------
    # Résultats
    # -------------------------

    print("Sommet :", (round(xs, 4), round(ys, 4)))

    print(
        "Intersections :",
        [round(i, 4) for i in intersections]
    )

    return intersections
```

```python
tracer_inegalite_valeur_absolue(1, -4, 80,-6, -1/5, -10, signe=">=", xmin=-30, xmax=50)
```
